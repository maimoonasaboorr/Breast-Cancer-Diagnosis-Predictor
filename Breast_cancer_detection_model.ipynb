Building "Usual Range" Hints for the App

For each of the 10 selected features, work out a friendly, rounded range (e.g. 0-40) close to but a bit wider than the actual data range, so it reads as a helpful guide rather than a strict data boundary.

```python
import math

def nice_upper_bound(x):
    """Round x up to a 'nice' looking number (40, 200, 3000, 0.3, ...)."""
    if x <= 0:
        return 1.0
    magnitude = 10 ** math.floor(math.log10(x))
    for m in [1, 1.5, 2, 3, 4, 5, 7, 10]:
        candidate = round(m * magnitude, 10)
        if candidate >= x:
            return float(candidate)
    return float(10 * magnitude)

def build_field_stats(df):
    stats={}
    for col in df.columns:
        col_min,col_max,col_median=df[col].min(),df[col].max(),df[col].median()
        value_range=max(col_max-col_min,1e-9)
        step=float(10**(math.floor(math.log10(value_range))-2))
        decimals=max(0,int(-math.floor(math.log10(step))))
        stats[col]={
            'min':0.0,
            'max':nice_upper_bound(col_max),
            'default':float(round(col_median,decimals)),
            'step':step,
            'decimals':decimals,
        }
    return stats

field_stats=build_field_stats(X_top)
field_stats
```
